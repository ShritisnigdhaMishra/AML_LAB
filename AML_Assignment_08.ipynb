```python
import numpy as np
x1=np.array([1,2,3,7,8,9])
x2=np.array([2,3,4,6,7,8])
x3=np.array([5,4,3,3,2,1])
y=np.array([0,0,0,1,1,1])
data={'x1':x1,'x2':x2,'x3':x3,'y':y}
def r(n,x,y):
  return abs((n*sum(x*y)-sum(x)*sum(y))/((n*sum(x**2)-sum(x)**2)*(n*sum(y**2)-sum(y)**2))**0.5)
def CFsMerit(k,x1,x2,y,n):
  rcf=(r(n,x1,y)+r(n,x2,y))/2
  rff=r(n,x1,x2)
  den_term=k+k*(k-1)*rff
  if den_term==0:
    return 0
  return (k*rcf)/(den_term**0.5)
def bestMerit(data):
  all_cols=list(data.keys())
  cols=[k for k in all_cols if k!="y"]
  feat_pairs=[(cols[i],cols[j])for i in range(len(cols))for j in range(i+1,len(cols))]
  n=len(data[cols[0]])
  res={}
  for x,y in feat_pairs:
    k=2
    res[(x,y)]=CFsMerit(k,data[x],data[y],data["y"],n)
  bestpair=max(res,key=res.get)
  return bestpair,res[bestpair]
bestpair,best_score=bestMerit(data)
print(f"The best CFs Merit Score is for {bestpair} with a score of {best_score:.4f}")
```
