```python
import pandas as pd
rainfall=['High','High','High','Low','Low','Low']
soiltype=['Loamy','Sandy','Clay','Loamy','Sandy','Clay']
yeild=['Good','Good','Good','Poor','Poor','Good']
def rone(df):
  col=list(df.columns)
  target=col[-1]
  length=len(df[target])
  resPer=None
  res=""
  for i in col[:-1]:
    currError=0
    currcol=set(df[i])
    for j in currcol:
      temp=df[df[i]==j][target].value_counts()
      if len(temp)>0:
        currError +=temp.sum()-temp.max()
    if resPer==None:
      resPer=currError/length
      res=i
    elif currError/length<resPer:
      resPer=currError/length
      res=i
  print(f"The feature '{res}' is selected with error rate of {resPer*100:.4f}%")
data={'Rainfall':rainfall,'Soiltype':soiltype,'Yield':yeild}
df=pd.DataFrame(data)
rone(df)
```
